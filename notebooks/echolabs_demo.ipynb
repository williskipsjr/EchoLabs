# EchoLabs: stem-aware AI mashups

Upload two songs and EchoLabs puts the **vocals of song A over the instrumental of song B**. It separates both songs with Demucs, matches their tempo **beat by beat**, and transposes the vocals to a compatible key.

* Code, method and benchmarks: https://github.com/williskipsjr/EchoLabs
* Listen to examples first: https://williskipsjr.github.io/EchoLabs/

**Tip:** choose *Runtime → Change runtime type → T4 GPU*. Separation is then roughly 10–20× faster than on CPU.

Only remix music you own or have permission to use (for example, Creative Commons tracks from ccMixter). Every mashup comes with a provenance file that credits its sources.
